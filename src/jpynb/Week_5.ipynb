```python
import random
```

```python
class Character:
    """
    Default class for Characters
    """
    def __init__(self,name,Hp,Mp,St,DX,IQ,LX):
        """
        Init: Generate Character as inputs.
        
        name: String
        Hp, Mp, ST, DX, IQ, LX: int
        """
        self.nickname=name
        self.HP=Hp; self.MP=Mp; self.ST=St
        self.DX=DX; self.IQ=IQ; self.LX=LX
    
    def acc(self):
        acc=random.randrange(5,11)/10+.01*self.LX
        return acc

    def attack(self,character):
        """
        Attack to character.
        
        character: Character
        """
        dm=(self.ST+self.DX+self.IQ)*self.acc()
        character.HP-=dm
    
    def status(self):
        """
        Get the general stats of the character.
        """
        if self.HP<0:
            print(f"{self.nickname}의 HP {self.HP}, MP {self.MP} : DIE")
        else:
            print(f"{self.nickname}의 HP {self.HP}, MP {self.MP}")
```

```python
class Warrior(Character):
    """
    Warrior class based on character
    """
    
    # not overriding acc(self): random.randrange(5,11)/10+.01*self.LX is same as the super()
    
    def attack(self,character):
        """
        Attack to character.
        Warrior: Strength has 4x of the stats about attacking
        
        character: Character
        """
        dm=(4*self.ST+self.DX+self.IQ)*self.acc()
        character.HP-=dm
    
    def skill(self,character):
        """
        Special attack of Warrior. Consumes 10 MP.
        
        character: Character
        """
        if self.MP>=10:
            dm=8*self.ST*self.acc()
            character.HP-=dm
        else: print("Not enough MP")
```

```python
class Archer(Character):
    """
    Archer class based on character
    """
    
    def acc(self):
        return random.randrange(7,11)/10+.1*self.LX
    
    def attack(self,character):
        """
        Attack to character.
        Archer: DEX has 2x of the stats about attacking
        
        character: Character
        """
        dm=(self.ST+2*self.DX+self.IQ)*self.acc()
        character-=dm
    
    def skill(self,character):
        """
        Special attack of Archor. Consumes 30 MP.
        
        character: Character
        """
        if self.MP>=30:
            dm=4*self.DX*self.acc()
            character.HP-=dm
            self.MP-=30
        else: print("Not enough MP")
```

```python
class Wizard(Character):
    """
    Wizard chass based on character
    """
    
    def acc(self): return random.randrange(3,11)/10+.3*self.LX
    
    def attack(self,character):
        """
        Attack to character.
        Wizard: Strength has 0x, DEX has 2x, Intelligence has 4x of the stats about attacking
        
        character: Character
        """
        dm=(2*self.DX+4*self.IQ)*self.acc()
        character.HP-=dm
    
    def skill(self,character):
        """
        Special attack of Wizard. Consumes 50 MP.
        
        character: Character
        """
        if self.MP>=50:
            dm=10*self.IQ*self.acc()
            character.HP-=dm
            self.MP-=50
        else: print("Not enough MP")
```

```python
# Create Characters
adv=Character('J',100,30,25,10,10,10)
war=Warrior('K',300,100,100,40,20,25)
arc=Archer('T',200,120,40,120,40,30)
wiz=Wizard('B',100,300,20,20,150,10)
```

```python
# 모험가가 공격
adv.attack(war) # assume adv.acc()=.7
adv.attack(arc)
adv.attack(wiz)

# 각각이 반격
war.skill(adv) # assume war.acc()=.9
arc.skill(adv) # assume arc.acc()=1.1
wiz.skill(war) # 넌 어디다 쏘니? # assume wiz.acc()=3.3

# 분노의 속공
for i in range(10): war.attack(wiz)

# 턴 종료, 상태 확인
adv.status()
war.status()
arc.status()
wiz.status()
```
