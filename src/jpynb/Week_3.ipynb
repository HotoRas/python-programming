```python
# append
thislist = ["apple", "banana", "cherry"]
thislist.append("orange")
print(thislist)
```

```python
# insert
thislist = ["apple", "banana", "cherry"]
thislist.insert(1, "orange")
print(thislist)
```

```python
# extend
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
thislist.extend(tropical)
print(thislist)
```

```python
# add any iterable
thislist = ["apple", "banana", "cherry"]
thistuple = ("kiwi", "orange")
thislist.extend(thistuple)
print(thislist)
```

```python
# remove item
thislist = ["apple", "banana", "cherry"]
thislist.remove("banana")
print(thislist)
```

```python
# remove the first occurance of specified
thislist = ["apple", "banana", "cherry", "banana", "kiwi"]
thislist.remove("banana")
print(thislist)
```

```python
# remove specified idx
thislist = ["apple", "banana", "cherry"]
thislist.pop(1)
print(thislist)
```

```python
# remove the last
thislist = ["apple", "banana", "cherry"]
thislist.pop()
print(thislist)
```

```python
# remove the 1st
thislist = ["apple", "banana", "cherry"]
del thislist[0]
print(thislist)
```

```python
# remove entire list
thislist = ["apple", "banana", "cherry"]
del thislist # remove variable thislist: list
print(thislist)# except error: thislist is removed so not defined
```

```python
# clear list
thislist = ["apple", "banana", "cherry"]
thislist.clear()
print(thislist)
```
