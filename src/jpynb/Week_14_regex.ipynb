```python
import re

data='''
park 800905-1049118
kim  700905-1059119
'''

result=[]
for line in data.split('\n'):
    word_result=[]
    for word in line.split('\t'):
        for temp in word.split(' '):
            if len(temp)==14 and temp[:6].isdigit() and temp[7:].isdigit():
                temp=temp[:6]+'-'+'*'*7
            word_result.append(temp)
    result.append(' '.join(word_result))

print('\n'.join(result))
```

```python
pat = re.compile(r'(\d{6})[-]\d{7}')
print(re.sub(pat,r'\g<1>-*******',data))
```

```python
p=re.compile('[a-z]+')
print(p.findall('life is too short'))
```

```python
p=re.compile(r'\d+')
print(p.findall('369 life is 222 short'))
```
