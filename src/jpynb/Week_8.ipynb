- 12자리의 비밀번호를 만들어야 함
- 첫 번째 문자는 대문자, 'O'는 제외
- 마지막 문자는 특수문자
- 숫자는 4개 이상
  - 소수는 1개 이상
- 특수 문자는 3개

특수문자 목록:
```py
'@#$%=:?./|~>*()<'
```

````python
import string, random
````

````python
numberList = string.digits # 0-9
uppercaseList = string.ascii_uppercase # A-Z
lowercaseList = string.ascii_lowercase # a-z
symbolList = ''.join(['@','#','$','%','=',':','?','.','/','|','~','>','*','(',')','<'])

pool=numberList+uppercaseList+lowercaseList+symbolList
````

````python
print(f"""{numberList}
{uppercaseList}
{lowercaseList}
{symbolList}
{pool}""")
````

````python
# test pool
print(''.join(random.sample(pool,12)))
````

````python
def passGen():
    while 1:
        pw = random.sample(pool,12)
        if pw[0] not in uppercaseList or pw[0]=='O':
            continue
        if pw[-1] not in symbolList:
            continue
        if len([i for i in pw if i in numberList])<4:
            continue
        if len([i for i in pw if i in '2357'])<1:
            continue
        if len([i for i in pw if i in symbolList])<3:
            continue
        return ''.join(pw)
````

````python
# get 10
for _ in range(10):
    print(f'password: {passGen()}')
````
