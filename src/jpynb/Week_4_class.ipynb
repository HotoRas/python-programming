```python
year=0
while year < 1900:
    year=int(input("연도?: "))

# 윤년 처리
if (lambda x: (x%4 == 0 and x%100 != 0 or x%400 == 0))(year):
    day_end=29
else:
    day_end=28

# 달력 출력
for month in range(1,13): # int: [1,13)
     for day in range(1,32): # int: [1,32)
         if month == 2: # 2월 별도 처리
             if day>day_end: # 28일까지, 윤년은 위에 처리한 대로
                 continue # 넘어가면 죽이기
         if month in [4,6,9,11]: # 30일인 달 별도 처리
             if day>30: # 넘어가면 죽이기
                 continue
         print(f"{year}년 {month}월 {day}일") # 일 단위 출력
     print("========================") # 월 단위 출력
```
