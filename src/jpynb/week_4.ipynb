```python
club=list(map(str,[1,2,3,4,5]))
# check: is it string?
print(club)
```

```python
# expand list in between specific idx
# to index 2
club[2:2]=['a','b','c']
# check
print(club)
```

```python
# issue: if using insert()
club.insert(2,['p','q','r'])
print(club)
```

```python
# extend can only insert elem. to the last pos
club.extend(['aa','bb','cc'])
print(club)
```

```python
# flush list in list element
def flatten(arg):
    ret=[]
    for i in arg:
        ret.extend(i) if isinstance(i,list) else ret.append(i)
    return ret
```

```python
print(flatten(club))
```

```python
boy=['김철수','나영석','강백호']
girl=['김연아','최윤희','이영애']
```

```python
# organize dictionary in step
dic={}
for (b,g) in zip(boy,girl): # zip() matches elements
    dic[b]=g

# check the content
print(dic)
```

```python
print(dic['나영석'])
```

```python
# dissolve dic w/ items() and re-organize
# f-string:
for (k,v) in dic.items():
    print(f'남자 {k}의 여자 파트너는 {v}이다.')
```
