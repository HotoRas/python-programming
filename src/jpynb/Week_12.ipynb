### Prompt 1

````python
def f1():
    print('연속적으로 실행시키는 프롬프트 제작')
    n=int(input('프롬프트 개수? '))
    promptList=[]
    for _ in range(n):
        prompt=input('프롬프트 내용? ')
        promptList.append(prompt)
    res='##'.join(promptList)
    #print('\n')
    #print(res)
    return res
````

### Prompt 2

````python
def f2():
    print('대상(예: 아인슈타인), 질문 (예: 상대성 이론)')
    n = int(input('질문 갯수? '))
    itemList=[]
    qList=[]
    
    for _ in range(n):
        item = input('질문할 대상? ')
        itemList.append(item)
        q = input(f'{item}에게 질문할 내용? ')
        qList.append(q)
    res = "\""+\
    f"itemList = {itemList}\n"+\
    f"qList = {qList}\n"+\
    "[ 당신은 <item>이야. <question>에 대해 설명해줘. for (item, question) in zip(<itemList>, <qList>)]"+"\""
    #print('\n')
    #print(res)
    return res
````

````python
def f3():
    print('특정 대상에 대해서 심도 있고 반복적인 질문하는 프롬프트 생성')
    n = int(input('아이템 갯수? ')) 
    itemList=[]
    
    for _ in range(n):
        item = input('질문할 대상? ')
        itemList.append(item)
    res="\""+\
          f"itemList={itemList}\n"+\
          '''
          ['<item>이 뭐야?'##
           '<item>은 뭐라고 할 수 있어?'##
           '<item>에 대해 누군가에게 설명한다면 뭐라고 할 거야?' for item in <itemList>]'''+"\"" 
    #print('\n')
    #print(res)
    return res
````

### Prompt 4

````python
def f4():
    print('관심 항목(예: 인공지능, 강아지, 머신러닝)를 대상(예: 유치원생,고등학생, 대학원생, 가정 주부)에게 설명')
    m=int(input('item 갯수? '))
    n=int(input('audience 갯수? '))
    
    itemList=[]
    audienceList=[]
    
    for _ in range(m):
        item = input('관심 item? ')
        itemList.append(item)
      
    for _ in range(n):
        audience = input('대상 audience? ')
        audienceList.append(audience)
        
    res="\""+\
          f"itemList={itemList}\n"+\
          f"audienceList={audienceList}\n"+\
    '''
    [너는 인공지능 <item>을 <audience>에게 소개해주는 강사야.
    뭐라고 소개해 줄래? for item in <itemList> for audience in <audienceList>]'''+"\"" 
    #print('\n')
    #print(res)
    return res
````

### Generator Preset Part (Global) ###


````python
# language setup
prompt_language = 'ko'
output_language = '한국어'

# base role (unused on Tree of Thoughts)
work_on = '구글'
work_role = '개발자'
    
# use ulfram alpha for output?
use_ulfram = False
    
# chain of thoughts, tree of thoughts
### when thought_tree is True, thought_chain will be ignored (assume as False)
### when both are False, it'll generate the basic one
thought_chain = False
thought_tree = True
tree_count = 5

# have the output? and which language?
do_output_code = False
code_out_lan = ['python','c#']

# have the chart?
### when do_output_map is True, do_output_flows is ignored (assume as False)
do_output_flows = False
do_output_map = False

# output difficulty
listener = '다른 개발자'

# modifier
temperature_p = 1
topP_p = 1
tone = 'friendly'
````

### Preset Generator

````python
def getPreset():
    """
    Creates preset using the given preset

    @returns: LLM preset prompt string
    """
    result = ''

    if thought_tree:
        result += f'''Imagine {tree_count} different experts are answering this <question>.
Every expert will write down one step of their thinking, then share it with the group.
Then all experts will go on to the next step, etc.
If any expert realises they're wrong at any point then they leave.
'''
    else:
        result += f'''너는 {work_on}의 {work_role}다.
'''
        if thought_chain:
            result += '한 단계 한 단계씩 생각해 봐라. 그리고 '
    
    result += f'{listener}에게 설명하듯이 답하라. '
    if do_output_code or do_output_map or do_output_flows:
        result += '그리고 '
        if do_output_code: 
            result += f'{", ".join(code_out_lan)} 코드'
            if do_output_map or do_output_flows:
                result += '와'
            result += ' '
        if do_output_map:
            result += '마인드맵'
        elif do_output_flows:
            result += '플로우 차트'
        result += '를 첨부해라.'
    result += f'\n{output_language}로 답해라.\n'

    result += f'''temperature: {temperature_p * 0.01}
top-p: {topP_p * 0.01}
tone: {tone}'''
    
    return result
````

````python
def printPreset():
    """
    Default prompt print function provided.

    @returns: the final preset, styled
    """
    preset_name = input('프리셋의 명칭을 지정해 주세요: ')

    preset = f'{preset_name}="{getPreset()}"'
    print(preset)
    return preset
````

### Prompt Generator

````python
def genPrompt():
    """
    LLM korean batch prompt generator

    @author: goqba

    @returns: large batch prompt usable for LLM
    """
    print('개별 프롬프트에 대한 다양한 미션(국문 번역, 문제 해결 코드 생성, 문제 풀이 순서도 작성 등)을 수행')

    item_count = int(input('item 개수? '))
    task_count = int(input('task 개수? '))

    itemList = []
    taskList = []

    for _ in range(item_count):
        item = input('관심 item? ')
        itemList.append(item)
    
    for _ in range(task_count):
        task = input('수행 task? ')
        taskList.append(task)

    result = '"'
    result += f'''itemList={itemList}
'''
    #for __ in range(int(item_count / 5)):
        #start = __ * 5
        #end = start + 5
        #result += f'''itemList{__}={itemList[start:end]}
#'''
    result += f'''taskList={taskList}

'''
    for _ in range(task_count):
        for __ in range(int(item_count / 5)):
            ## batch prompt runtime timeout fix
            #result += f'''[<item>에 대해 <task[{_}]>을 수행해 줘. for item in <itemList{__}>]
#'''
            result += f'''[<item>에 대해 <task>을 수행해 줘. for item in <itemList> for task in <taskList>]
'''
            ## batch prompt runtime timeout fix
            #if not (_ == task_count - 1 and __ == int(item_count / 5) - 1):
                #result += '##\n'
    
    return result
````

````python
def printPrompt():
    """
    Just prints the batch prompt generated with function: genPrompt()

    @returns: the prompt, for sure
    """
    prompt = genPrompt()
    print(prompt)

    return prompt
````
