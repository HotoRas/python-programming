````python
import random
````

````python
def isPrime(n):
    """소수를 확인한다. n: number (int)"""
    if n<2:
        return False
    for i in range(2, int(n**.5)+1): # sqrt(n)+1 까지만 확인하면 됨
        if n%i == 0:
            return False
    return True
````

````python
# num_list=range(1,46)

# generate Prime lists: drop 13
prime_list=[i for i in range(1,46) if isPrime(i) and i!=13]
````

````python
def genlotto():
    """로또 번호를 규칙에 따라 추첨한다."""
    while True:
        lotto=sorted(random.sample(range(1,46),6))
        
        if 13 in lotto: # 조건 1. 13은 나오면 안 됨
            continue
        
        if len([i for i in lotto if i in prime_list]) < 2: # 조건 2. 소수는 적어도 2개 이상
            continue
        
        if max(lotto)-min(lotto)>30: # 조건 3. 최대 수와 최소 수의 차이는 30을 넘어서는 안 됨
            continue
        
        if sum(lotto)/6<15 or sum(lotto)/6>20: # 조건 4. 나온 수의 평균은 15와 20 사이에 존재해야 함
            continue
        
        return lotto
````

````python
for _ in range(3):
    print(genlotto())
````
