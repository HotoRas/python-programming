## Regex

```python
import re
```

```python
# string starting with t and have single ltr
text = 'In the face of ambiguity, refuse the temptation to guess.'
match_list = re.findall(r't.',text)

print(match_list)
```

```python
# spacing to underscore
text = 'Beautiful is better than ugly.'
replaced = re.sub(r'\s','_',text) # replace
print(replaced)
```

```python
# split with ltrs wo intager, alphabet
text = "Special cases aren't special enough to break the rules."
splitted = re.split(r'[^a-zA-Z0-9]+',text) # split
print(splitted)
```

## String function for DNA

```python
def dna_counter(dna: str):
    count_a,count_c,count_g,count_t = 0,0,0,0
    for base in dna:
        if base=='A': count_a += 1
        elif base=='C': count_c += 1
        elif base=='G': count_g += 1
        elif base=='T': count_t += 1
    return (count_a,count_c,count_g,count_t)
```

```python
print(dna_counter(''))
```

```python
print(dna_counter('ACCGGGTTTT'))
```

```python
print(dna_counter('123XYZ'))
```

```python
def dna_counter_dic(dna: str): # dictionary
    counts={}
    for base in dna:
        if base not in counts: counts[base]=0
        counts[base]+=1
    return (counts.get('A',0),counts.get('C',0),counts.get('G',0),counts.get('T',0))
```

```python
[dna_counter_dic(seq) for seq in ['','ACCGGGTTTT','123XYZ']]
```

```python
def dna_counter_dic2(dna: str):
    counts={'A':0,'C':0,'G':0,'T':0,} # reset dict
    for base in dna:
        counts[base] += 1
    return (counts.get('A',0),counts.get('C',0),counts.get('G',0),counts.get('T',0))
```

```python
dna_counter_dic2('ACCGGGTTTT') # if not in list('ACGT') it throws error
```
