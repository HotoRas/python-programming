```python
# 정형화하기
def gugu(n: int, target: int = 9, step = 1) -> list:
    result = []
    for i in range(1,target + 1, step):
        result.append(f'{n}*{i}=={n*i}\n')
    return result

# 근데 익명함수로 되냐 이거? (됐길래 건너뛰기도 넣어봄)
#gugu = lambda n, target=9, step=1: [i*n for i in range(1,target+1,step)] # 되네??

# 바꾼거도 람다로
#gugu = lamdda n, target=9, step=1: [f'{n} * {i} == {n*i} for i in range(i,target+1,step)]
```

```python
# 제곱도 해보자구
#def mypower(n):
#    result = []
#    for i in range(1,11):
#        result.append(f'{n} ** {i} == {n**i}\n')
#    return result
# 얘도 람다
powr=lambda n,target=10,step=1: [f'{n} ** {i} == {n**i}\n' for i in range(1,target+1,step)]
```

```python
print(gugu(3))
print(gugu(19))
print(gugu(19,19))
print(gugu(19,19,2))
## 왜 되냐
```

```python
# 구구단 표를 만들자, 일단 뽑아보고
for i in range(2,4):
    print(''.join(gugu(i,9)))
    print('-'*100)
```

```python
for i in range(2,4):
    print(''.join(powr(i)))
    print('-'*50)
```

```python
for i in range(15,20):
    print(''.join(powr(i,19)))
    print('-'*40)
```

```python
# 총 페이지를 구해보자
class Paging:
    ## 여기서도 람다 쓰는 미친놈
    get_total_page=lambda self,content,content_per_page: content//content_per_page if content%content_per_page==0 else content//content_per_page+1
```

```python
server = Paging()
print(server.get_total_page(5,10))
print(server.get_total_page(30,10))
```
