```python
import re # regex lib.
```

```python
# search dials 1 (re.search())
p = re.compile(r'\d+\-\d+\-\d+') # "\-" == [-]
text = 'My number is 415-555-4242 or 010-1234-5678'
res=re.search(p,text)

print('Phone number found: '+res.group())
```

```python
# search dials 2 (re.findall() -> list)
p = re.compile(r'\d+\-\d+\-\d+') # "\-" == [-]
text = 'My number is 415-555-4242 or 010-1234-5678'
res=re.findall(p,text)

print('Phone number found: '+res[0])
```

```python
# string lookup
p = re.compile(r'Bat(man|mobile|copter)') # 'Bat' + contents in () will make match, and things in () will be group
text = 'Batman uses a Batmobile rather than a Batcopter'
res = re.findall(p,text)

## add 'Bat' to every result since it's excluded
res = ['Bat' + match for match in res]
### if you don't want it: use r'Bat(?:man|mobile|copter)' as regex
print(res)
```

```python
# string lookup: complecated
# p = re.compile(r'Batman|Batwoman') # why?
p = re.compile(r'Bat(?:wo)?man') # flagging ?: at the very first inside () will make group not exported
text = 'Batman loves Batwoman'
res = re.findall(p,text)
print(res)
```

```python
# string combined with intager and spaces
p = re.compile(r'\d+\s\w+') # some intagers, single spaces, some words in [a-zA-Z_] in a row.
text = '12 drummers, 11 pipers, 10 lords, 9 ladies, 8 maids, 7 swans, 6 geese,\
    5 rings, 4 birds, 3 hens, 2 doves, 1 paetridge'
res = re.findall(p,text)
print(res)
```

```python
# AEIOU
p = re.compile(r'[aeiouAEIOU]')
text = 'RoboCop eats baby food. BABY FOOD.'
res = re.findall(p,text)
print(res)
```

```python
# then if not?
p = re.compile(r'[^aeiouAEIOU]')
text = 'RoboCop eats baby food. BABY FOOD.' # using same string
res = re.findall(p,text)
print(res)
```

```python
# now exclude spaces too
p = re.compile(r'[^aeiouAEIOU\s]')
text = 'RoboCop eats baby food. BABY FOOD.'
res = re.findall(p,text)
print(res)
```

```python
# Wildcards
p = re.compile(r'.at')
text = 'The cat in the hat sat on the flat mat.'
res = re.findall(p,text)
print(res)
```

```python
# match all (using .*)
p = re.compile(r'First Name: (.*) Last Name: (.*)') # things in () will match after things outside, even if greedy
text = 'First Name: AI Last Name: Sweigart'
res = re.findall(p,text)
print(res)
```

```python
# ignore caps
p = re.compile(r'robocop\s\d+',re.I) # re.I (caps i) will ignore caps
text = 'Robocop 1, ROBOCOP 2, rObOcOP 3, roboCOP 4, RoBoCoP 5'
res = re.findall(p,text)
print(res)
```

```python
# * but not not greedy
p = re.compile(r'<.*?>')
text = '<To serve man> for dinner.>'
res = re.findall(p,text)
print(res)
```

```python
# if greedy:
p = re.compile(r'<.*>')
text = '<To serve man> for dinner.>' # using text above
res = re.findall(p,text)
print(res)
```

```python
# Replace some
p = re.compile(r'Agent \w+') # matches 'Agent ' and some words in a row
text = 'Agent Alice gave the secret documents to Agent Bob.'
res = re.sub(p, 'CENSORED', text) # compiled regex, replacing word, source -> string
print(res)
```

```python
# dials
################################################
# regex explain                                #
################################################
# ((area code?)(sep?)(first)(sep)(last)(ext?)) #
################################################
p=re.compile(r'''(
(\(?0\d{2}\)?)? # 국번 (이때 모든 국번은 0으로 시작함)
(\s|-|\.)? # sep
\d{4} # 앞 4자리
(\s|-|\.) # sep
\d{4} # 뒤 4자리
(\s*(ext|x|ext.)\s*d{2,5})? # extension
)''', re.VERBOSE)

text = '(082)-5900-7777, 010-1234-5678, 010 3333 8888'

res = re.findall(p,text)
print(list(map(lambda x: x[0],res)))
```

```python
# 재미로, 국제전화 함 해볼게요

p = re.compile(r'''(
(\(?(
(\+\d{1,3})? # 국제전화 확장
(\s|-|\.)? # sep
|0 # 또는 0
)\d{2}\)?)? # 국번 (국제전화 예외처리)
(\s|-|\.)? # sep
\d{4} # 앞 4자리
(\s|-|\.) # sep
\d{4} # 뒤 4자리
(\s*(ext|x|ext.)\s*d{2,5})? # extension
)''', re.VERBOSE)

text = '''
+82-10-1254-8854
010 4853 4865
025 1487-4572
(010) 2258 7744
(082)-5900-7777, 010-1234-5678, 010 3333 8888
'''

res = re.findall(p,text)
print(list(map(lambda x: x[0],res)))
```
