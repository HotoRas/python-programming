```python
class Calculator:
    def __init__(self):
        self.result=0
    
    def add(self,num):
        self.result+=num
        return result
    
    def sub(self,num):
        self.result-=num
        return result
    
    def mul(self,num):
        self.result*=num
        return num
    
    def div(self,num):
        self.result/=num
        return num
    
    def mem_reset(self):
        self.__init__()
```

```python
calc1=Calculator()
print(calc1.add(3))
print(calc1.sub(1))
print(calc1.mul(37))
print(calc1.div(24))
calc1.mem_reset()
print(calc1.add(0))
```
