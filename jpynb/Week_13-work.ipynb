````python
'''
그냥 귀찮아서 만들어보는 파이썬 파일
근데 이거 만드는 게 더 귀찮아 보임
'''

def _getAnswer(_type: str) -> str:
    print(r'받아올 변수를 입력할 때에는 <변수명>으로 입력하세요')
    line = 0
    out = ''
    while 1:
        print(f'현재 줄 수: {line}\n입력된 전체 {_type}:\n{out}')
        print()
        line += 1
        out += input(f'{_type} ({line}번째 줄): ')
        out += '\n'
        print()
        isStop = ''
        print()
        print(f'현재 줄 수: {line}\n입력된 전체 {_type}:\n{out}')
        print()
        while not (isStop.lower() == 'y' or isStop.lower() == 'n'):
            isStop = input(f'{_type} 입력을 종료하시겠습니까? (y: 예, n: 아니오): ')
        if isStop.lower() == 'y': return out
````

````python
def _getTask() -> list:
    print(r'작업을 받아와서 리턴합니다.')
    out=[]
    while 1:
        print(f'현재 작업 수: {len(out)}\n전체 목록: {out}')
        print()
        out_apd = input(f'추가할 작업 입력 (현재 {len(out)}개): ')
        out.append(out_apd)
        isStop = ''
        print()
        print(f'현재 작업 수: {len(out)}\n전체 목록: {out}')
        print()
        while not (isStop.lower() == 'y' or isStop.lower() == 'n'):
            isStop = input('작업 입력을 종료하시겠습니까? (y: 예, n: 아니오): ')
        if isStop.lower() == 'y': return out
````

````python
def prompt_with_limits_and_format() -> str:
    print('질문, 제한 사항, 포맷 및 그 예제를 제공하고 코드를 제공받는 프롬프트')

    language_out_standard = input('출력받을 언어: ')
    language_out_var = input('출력받을 언어의 변수: ')
    
    question = _getAnswer('질문')
    limit = _getAnswer('제한 사항')
    form = _getAnswer('포맷')
    form_example = _getAnswer('포맷 설명')
    
    out = f'''"
{language_out_var}='{language_out_standard}'

# 문제
{question}
# 제한 사항
{limit}
# 입출력 예
{form}
# 입출력 예 설명
{form_example}
"'''
    return out
````

````python
def prompt_request_batch() -> str:
    print('사용자 제공되는 코드에 대해 작업을 수행하도록 하는 프롬프트')
    code_lang = input('코드의 언어 단축명 입력 (예: py, py3, python, cpp...): ')
    code = _getAnswer('코드')
    task = _getTask()
    out=f'''"
code='
```{code_lang}
{code}```

taskList=['{"','".join(task)}']
[<code>에 대해 <task>를 수행해줘. for task in <taskList>]
"'''
    return out
````

````python
prompt=prompt_with_limits_and_format()
print(prompt)
````

````python
prompt=prompt_request_batch()
print(prompt)
````
