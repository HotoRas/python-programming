```python
## 11번
def isPrime(number):
    if number<2:
        return False
    for i in range(2,int(number**.5)+1):
        if number%i==0:
            return False
    return True
```

```python
## 17번
'''
김 대리는 업무에 필요한 물건들을 구매하기 위해 현금을 받아 문구점에 가는 길입니다.
물건들을 구입한 후 남은 잔액은 다시 반납해야 합니다. 예를 들어 구매한 물건의 가격이 [2100, 3200,2100,800]이면
총 구매금액은 8,200원입니다. 지급 받은 현금이 10,000원이라면 1,800원을 반납하면 됩니다.
김대리가 구매한 물품들의 가격이 들어있는 리스트 price와 받은 금액 money가 매개변수로 전달될 때, 반납해야 하는
금액을 return 하도록 solution 함수를 완성하시오,
단, 모자라는 경우 -1을 반환하도록 하시오.
'''
```

```python
def solution(price,money):
    ans=0
    for i in price:
        money-=i
    ans=money
    if ans<0:
        return -1
    return ans
```

```python
## 18번
'''
리듬체조 선수 영심은 지난 1년간 대회에 참가한 기록을 정리 중이다. 각 대회에서 받은 점수가 들어있는
리스트가 주어졌을 때, 이 선수의 최고 점수와 최저 점수를 제외한 나머지 점수들의 합계를 구하려 한다.
이를 위해 다음의 단계로 프로그램 구조를 작성했다.
1. 모든 점수의 합을 구한다.
2. 최고 점수와 최저 점수를 구한다.
3. (모든 점수의 합)-(최고 점수)-(최저 점수)의 값을 return 한다.

대회별 점수가 들어있는 리스트 scores가 매개변수로 전달될 때, 최고 점수와 최저 점수를 제외한
나머지 점수의 합을 return 하도록 solution 함수를 작성하시오
'''
```

```python
def getMaxMin(s):
    s.sort()
    return s[-1],s[0]

def solution2(scores):
    tot=sum(scores)
    mx,mn=getMaxMin(scores)
    return tot-mx-mn
```

```python
## 19번
'''
A 고등학교에서는 수준별 수능 대비반을 운영하려고 한다. 용기반은 최근 수능 모의고사에서 200점 이하의 성적을 취득한
학생만을 대상으로 한다. 학생들의 모의고사 성적이 들어있는 리스트 scores가 매개변수로 전달될 때, 용기반 대상자의
인원수를 return하는 함수를 작성하시오.
[주] 점수는 0 이상 400 이하의 정수이다.
'''
```

```python
def solution3(scores):
    cnt=0
    for _ in scores:
        if _ <= 200:
            cnt+=1
    return cnt
```

```python
## 20번
'''
율무국에서는 국가 신용 등급에 따라 수입하는 물품에 관세를 부여한다. 국가 신용 등급에 따른
관세율은 다음과 같다. (S=개발도상국, G=신진부흥국, V=부흥국)
------------------------------------------------------------
등급 관세율
------------------------------------------------------------
S 5%
------------------------------------------------------------
G 10%
------------------------------------------------------------
V 15%
------------------------------------------------------------
예들 들어 등급이 'S'인 국가에서 수입하는 물품의 가격이 1,000원인 경우 관세율 5%를 적용한
가격은 1,050원이 된다. 물품의 가격 price와 원산지국가의 신용등급을 나타내는 문자열 grade가
매개변수로 주어질 때, 관세율을 적용한 물품 가격을 return하는 solution 함수를 작성하시오.
[주] 모든 가격은 소수점 이하를 버린 정수만 사용한다.
'''
```

```python
def solution4(price,grade):
    ans=0
    if grade=='S': # startup
        ans=price*1.05
    if grade=='G': # growth
        ans=price*1.1
    if grade=='V': # 뭐더라이거
        ans=price*1.15
    return int(ans) # round-down
```

```python
## 21번
'''
369 게임은 순차적으로 숫자를 부르면 일정 규칙에 따라 숫자 대신 박수를 치는 게임이다.
게임의 규칙은 다음과 같다. 1부터 시작하고 한 사람씩 차례대로 1씩 더해가며 말한다.
말해야 하는 숫자에 3,6,9가 포함되어 있으면 숫자 대신 포함되어 있는 갯수만큼 손뼉을
쳐야 한다. 예를 들어 1부터 12까지 게임을 진행하는 경우 손뼉을 치는 경우은 다음과 같다.
1 2 짝 4 5 짝 7 8 짝 10 11 12
손뼉은 총 3번을 치게 된다. 어떤 수 number가 매개변수로 전달될 때, 1부터 number까지 연속하는 숫자들에 대해
369 게임을 올바르게 진행했을 경우 박수를 총 몇 번 치게 되는지를 return 하는 solution 함수를 작성하시오.
[주] 전달되는 number는 2자리 이상의 정수이다.
'''
```

```python
def g369Claps_sol(num):
    cnt=0
    for i in range(1,num+1):
        cur=i
        while cur!=0:
            unit=cur%10
            if unit in '369':
                cnt+=1
            cur//=10
    return cnt
```
