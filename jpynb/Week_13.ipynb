## try-else

```python
try:
    age=int(input('나이를 입력하세요: '))
except:
    print('입력이 정확하지 않습니다.')
else:
    if age <= 18: print('미성년자는 출입 금지입니다.')
    else: print('환영합니다.')
```

```python
### source: https://wikidocs.net/193066 ###

# 함수 chk_dup_numbers 정의. 이 함수는 입력받은 문자열 s에 중복되는 숫자가 있는지 검사하고, 
# 그 결과를 Boolean 값으로 반환합니다.
def chk_dup_numbers(s):
    result = []  # 중복되지 않은 숫자를 저장할 빈 리스트 생성

    # 문자열 s에 대해 반복문 실행
    for num in s:
        # 현재 숫자가 result 리스트에 없다면
        if num not in result:
            result.append(num)  # 리스트에 추가
        else:
            return False  # 중복된 숫자가 발견되면 False 반환

    # 모든 숫자가 중복되지 않고, 그 개수가 10개라면 True 반환. 그렇지 않다면 False 반환.
    return len(result) == 10
```
