```python
# closure, decorator
def mul(m: int):
    def wrapper(n: int)->int:
        return m*n
    return wrapper

print(mul(3)(5))

import time
def elapsed(_origin):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=_origin(*args,**kwargs)
        end=time.time()
        print(f'elapsed time: {end-start} s')
        return result
    return wrapper

@elapsed
def dummy(msg: object)->None:
    print(f'printing "{msg}"')
    return

dummy('yo hello how are you')
```

```python
# iterator, generator
a=list(range(100000))
ia=iter(a)
print(next(ia))
print(next(ia))
```

```python
a=range(1000000000000000000000)
for _ in iter(a):
    print(_,end=' ')
    if _>100: break
```

```python
class myIter:
    def __init__(self,data):
        self.data=data
        self.position=0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.position>=len(self.data): raise StopIteration
        result = self.data[self.position]
        self.position+=1
        return result
```

```python
i=myIter(range(100000000000000000))
for item in i:
    print(item)
    if item>100: break
```

```python
def myGen():
    yield 'a'
    yield 'b'
    yield 'c'
    yield 'd'
    yield 'e'
    yield r'너 탈당 좀 해라 준슥아' # 엄
```

```python
g=myGen()
print(type(g))
```
